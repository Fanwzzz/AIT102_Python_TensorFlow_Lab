#### Exercise 3


```python
time = int(input("Please enter number of seconds:"))
days = time // (60*60*24)
hours = (time%(60*60*24)) // (60*60)
minutes = (time%(60*60)) // 60
seconds = time % 60
print(f"{time} seconds are equivalent to {days} days, {hours} hours, {minutes} minutes and {seconds} seconds")
```

#### Exercise 4


```python
num = int(input("Please enter a whole number:"))
squared_value = num ** 2
remainder = num % 2
print(f"The square of {num} is {squared_value}")
print(f"The remainder when divided by 2 is {remainder}")
if remainder == 0:
    print(f"{num} is even")
else:
    print(f"{num} is odd")
```

#### Exercise 5


```python
total_items = int(input("Please enter total items:"))
number_of_groups = int(input("Please enter number of groups:"))
standard_result = total_items / number_of_groups
floor_result = total_items // number_of_groups
print(f"Standard division result is {standard_result}")
print(f"Floor division result is {floor_result}")
```

#### Exercise 6


```python
items = []
item1 = input("Please enter first item:")
item2 = input("Please enter second item:")
item3 = input("Please enter third item:")
items.append(item1)
items.append(item2)
items.append(item3)
print("The original list is", items)
item2 = input("Please enter a new second item: ")
items[1] = item2
print("The updated list is", items)
print("The length of the list is", len(items))
```
